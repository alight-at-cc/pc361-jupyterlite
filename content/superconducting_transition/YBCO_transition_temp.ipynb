# Superconducting Phase Transition in YBCO
The transition from a metallic to a superconducting state is typically what's called a second-order or "continuous" phase transition.  The resistance drops to zero quickly and continuously as the temperature is brought down across the superconducting transition or "critical" temperature, $T_c$.

In this exercise, your task is to determine this transition temperature and its uncertainty from a set of resistance versus temperature measurements.  Here's a few bits of setup to get you started:

```python
import numpy as np #label namespace of Numpy functions as 'np' for shorter commands
import matplotlib.pyplot as plt #same for simple plotting

#load the labels and uncertainties from the first line of your group's data file
#  open the file by creating a file object with the name 'data_file'
data_file = open("conductivity_data_group_x.txt", 'r')
#  read the first line
column_label = data_file.readline()
#  look at the text in the first line, which contains the column label(s)
print(column_label)
```

Now, let's break the data into separate variables so that we can work with them more easily.  I've used the data labeled 'group x' but you should use the set you've been assigned.  

```python
#so the first column is temperature and the second is resistance
#label the uncertainties
delta_T = 0.5 #K
delat_R = 0.1 #Ohms

#load the resistance data from your group's data file into a numpy array for analysis
# skip the first row that has the column labels
data = np.loadtxt("conductivity_data_group_x.txt",delimiter=',',skiprows=1) 
#  check that the data looks like the numbers in the text file by printing a few rows
print(data[0:5,:])

#we can split it into two variables for easier manipulation
T = data[:,0] #first column (K)
R = data[:,1] #second column (Ohms)
```
